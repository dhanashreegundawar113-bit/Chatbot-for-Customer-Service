```python

# Frequently asked questions
faqs = [
    {
        "question": "What are your business hours?",
        "answer": "Our customer service is available Monday to Saturday, from 9 AM to 6 PM."
    },
    {
        "question": "How can I track my order?",
        "answer": "You can track your order using the tracking link provided in your order confirmation email."
    },
    {
        "question": "What is your return policy?",
        "answer": "You can request a return according to the return policy and eligibility conditions of the store."
    },
    {
        "question": "How can I cancel my order?",
        "answer": "Please visit your order history and check whether the order is eligible for cancellation."
    },
    {
        "question": "What payment methods do you accept?",
        "answer": "Available payment methods may include UPI, debit cards, credit cards, and other options shown at checkout."
    },
    {
        "question": "How long does delivery take?",
        "answer": "Delivery time depends on your location and the shipping option selected during checkout."
    },
    {
        "question": "How can I contact customer support?",
        "answer": "Please use the contact details or support form provided on the company's official website."
    },
    {
        "question": "I forgot my password. What should I do?",
        "answer": "Select 'Forgot Password' on the login page and follow the instructions to reset your password."
    },
    {
        "question": "How do I get a refund?",
        "answer": "If your order is eligible, request a refund through the store's official return or support process."
    },
    {
        "question": "Thank you",
        "answer": "You're welcome! I'm happy to help."
    }
]

print("FAQ dataset created successfully!")
print("Total FAQs:", len(faqs))
```

```python

# Extract questions from the dataset
questions = [faq["question"] for faq in faqs]

# Convert text into numerical vectors
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

faq_vectors = vectorizer.fit_transform(questions)

print("NLP model trained successfully!")
```

```python

def chatbot_response(user_input):
    user_input = user_input.strip()

    if not user_input:
        return "Please enter a question."

    # Handle greetings
    greeting_words = ["hi", "hello", "hey"]
    cleaned_input = re.sub(r"[^\w\s]", "", user_input.lower())

    if cleaned_input in greeting_words:
        return "Hello! Welcome to Customer Support. How can I help you?"

    # Convert the user's question into a vector
    user_vector = vectorizer.transform([user_input])

    # Calculate similarity with all FAQ questions
    similarity_scores = cosine_similarity(
        user_vector, faq_vectors
    ).flatten()

    best_match = similarity_scores.argmax()
    best_score = similarity_scores[best_match]

    # Avoid returning unrelated answers
    if best_score < 0.15:
        return (
            "Sorry, I couldn't understand your question. "
            "Please rephrase it or contact customer support."
        )

    return faqs[best_match]["answer"]


print("Chatbot is ready!")
```

```python

print("================================")
print("   CUSTOMER SERVICE CHATBOT")
print("================================")
print("Type 'exit' to end the conversation.\n")

while True:
    user_message = input("You: ")

    if user_message.lower().strip() == "exit":
        print("Bot: Thank you for contacting us. Goodbye!")
        break

    response = chatbot_response(user_message)
    print("Bot:", response)
    print()
```
